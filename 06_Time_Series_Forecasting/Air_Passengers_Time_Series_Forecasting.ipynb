```python
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.arima.model import ARIMA

# Load dataset
df = pd.read_csv('AirPassengers (1).csv')

# Date conversion
df['Month'] = pd.to_datetime(df['Month'])
df.set_index('Month', inplace=True)

# Time Series Visualization
plt.figure(figsize=(10,5))
plt.plot(df['#Passengers'])
plt.title('Monthly Airline Passengers')
plt.xlabel('Month')
plt.ylabel('Passengers')
plt.show()

# Moving Average
df['Moving_Average'] = df['#Passengers'].rolling(12).mean()

plt.plot(df['#Passengers'], label='Original')
plt.plot(df['Moving_Average'], label='Moving Average')
plt.legend()
plt.show()

# ARIMA Model
model = ARIMA(df['#Passengers'], order=(2,1,2))
model_fit = model.fit()

# Forecast
forecast = model_fit.forecast(steps=12)

print("Next 12 Months Prediction:")
print(forecast)

# Forecast Graph
plt.plot(df['#Passengers'], label='Actual')
plt.plot(forecast, color='red', label='Forecast')
plt.title('Passenger Forecast')
plt.legend()
plt.show()
```
